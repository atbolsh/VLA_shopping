# ChatVLA-1 — official evaluate, same weights

Kernel **chatvla**. ChatVLA-2 robot weights are **not** public.

Three panes are the **same** EO-1 `example1.jpg` stand-in. Not their ALOHA cameras.
Success = `evaluate` `outputs` contain English **and** an action.

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
sys.path.insert(0, str(ROOT / "src"))
from IPython.display import display, Markdown
from PIL import Image
from chatvla_turns import ChatVLAPlay, OFFICIAL_TASK
SHOT = ROOT / "screenshot"
top = Image.open(SHOT / "top.jpg").convert("RGB")
left = Image.open(SHOT / "left_wrist.jpg").convert("RGB")
right = Image.open(SHOT / "right_wrist.jpg").convert("RGB")
display(Markdown("**What it sees** (one stand-in, three slots)"))
display(top)
```

```python
play = ChatVLAPlay()
print("loaded", play.weights, "err", play.err)
out = play.evaluate(top, left, right, OFFICIAL_TASK, eval_in_vqa=False)
display(Markdown("**Raw outputs**"))
print(out["text"] or "_(empty)_")
display(Markdown(f"**Verdict** `{out['verdict']}` · `{out['path']}`"))
display(Markdown("**Action**"))
print(out["action"])
```
