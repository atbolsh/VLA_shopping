# EO-1 — same 3B, text + action

Kernel **eo1**. Official `processor.generate`. No second VLM.

3B Qwen2.5-VL — below the 12B bar. Success = readable English **and** an action array.

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
sys.path.insert(0, str(ROOT / "src"))
from IPython.display import display, Markdown
from PIL import Image
from eo1_turns import EO1Play, OFFICIAL_TASK
SHOT = ROOT / "screenshot"
img_p = SHOT / "example1.jpg"
wrist_p = SHOT / "example2.png"
if not img_p.is_file():
    raise FileNotFoundError("screenshot/ empty. In the eo1 venv: python download_frames.py")
img = Image.open(img_p).convert("RGB")
wrist = Image.open(wrist_p).convert("RGB") if wrist_p.is_file() else img
display(Markdown(f"**What it sees** `{img_p.name}` {img.size} / wrist `{wrist_p.name}`"))
display(img)
display(wrist)
```

```python
play = EO1Play()
print("loaded", play.weights)
out = play.generate(img, wrist, OFFICIAL_TASK)
display(Markdown("**Raw text** (same weights)"))
print(out["text"] or "_(empty)_")
display(Markdown(f"**Verdict** `{out['verdict']}` · path `{out['path']}`"))
display(Markdown("**Action**"))
print(out["action"])
```
