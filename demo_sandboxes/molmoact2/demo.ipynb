# MolmoAct2-Think + Molmo2-ER

Two official surfaces. Kernel **molmoact2**. Do not feed Gemma-style chat into `predict_action`.

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
sys.path.insert(0, str(ROOT / "src"))
from IPython.display import display, Markdown
from PIL import Image
from think_act import ThinkLibero
from molmo2_ask import Molmo2Ask

third = Image.open(ROOT / "assets" / "libero_third.png").convert("RGB")
wrist = Image.open(ROOT / "assets" / "libero_wrist.png").convert("RGB")
display(Markdown("**What it sees**"))
display(third)
display(wrist)
```

## Act / think — `predict_action` + adaptive depth

```python
thinker = ThinkLibero()
task = "pick up the black bowl and place it on the plate"
act = thinker.act(third, wrist, task)
display(Markdown("**Think (depth tokens / updated cells)**"))
print(act["think"])
display(Markdown("**Reply (actions)**"))
display(Markdown(act["reply"]))
```

## Ask — Molmo2-ER chat template (separate load)

```python
asker = Molmo2Ask()
ask = asker.ask(third, "What objects are on the table, and what should the arm do first?")
display(Markdown("**Reply**"))
display(Markdown(ask["reply"]))
```
