# MolmoAct2-Think + Molmo2-ER

Two official surfaces. Kernel **molmoact2**. Do not feed Gemma-style chat into `predict_action`.

Frames in `screenshot/` are the Think-LIBERO card samples (`libero_10` / episode 0 / frame 0). If that folder is empty:

`source .venv/bin/activate && python download_frames.py`

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
sys.path.insert(0, str(ROOT / "src"))
from IPython.display import display, Markdown
from PIL import Image
from think_act import ThinkLibero, OFFICIAL_TASK
from molmo2_ask import Molmo2Ask

SHOT = ROOT / "screenshot"
third_p = SHOT / "sample_agentview_rgb.png"
wrist_p = SHOT / "sample_wrist_rgb.png"
if not third_p.is_file() or not wrist_p.is_file():
    raise FileNotFoundError(
        "screenshot/ missing official frames. In the molmoact2 venv: python download_frames.py"
    )
third = Image.open(third_p).convert("RGB")
wrist = Image.open(wrist_p).convert("RGB")
display(Markdown("**What it sees** (official Think-LIBERO sample, not the placeholder squares)"))
display(Markdown(f"agentview `{third_p.name}` {third.size}"))
display(third)
display(Markdown(f"wrist `{wrist_p.name}` {wrist.size}"))
display(wrist)
```

## Act / think — `predict_action` + adaptive depth

Task and 8-D EEF state are the ones on the [Think-LIBERO card](https://huggingface.co/allenai/MolmoAct2-Think-LIBERO). Restart the kernel after pulling `src/think_act.py`.

```python
thinker = ThinkLibero()
task = OFFICIAL_TASK
act = thinker.act(third, wrist, task)
display(Markdown("**Think (depth bins / updated cells)**"))
print(act["think"])
display(Markdown("**Reply (actions)**"))
print(act["reply"])
```

## Ask — Molmo2-ER chat template (separate load)

```python
asker = Molmo2Ask()
ask = asker.ask(third, "What objects are on the table, and what should the arm do first?")
display(Markdown("**Reply**"))
print(ask["reply"] or "(empty)")
```
