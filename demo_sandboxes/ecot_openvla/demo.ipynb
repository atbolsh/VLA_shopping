```python
from pathlib import Path
import sys
ROOT = Path.cwd()
sys.path.insert(0, str(ROOT / "src"))
from IPython.display import display, Markdown
from PIL import Image
from ecot_turns import ECoTPlay, OFFICIAL_TASK
img_p = ROOT / "screenshot" / "test_obs.png"
if not img_p.is_file():
    raise FileNotFoundError("screenshot/test_obs.png missing. python download_frames.py")
img = Image.open(img_p).convert("RGB")
display(Markdown(f"**What it sees** `{img_p.name}` {img.size} (official test_obs.png)"))
display(img)
```

```python
play = ECoTPlay()
print("loaded", play.weights)
out = play.predict(img, OFFICIAL_TASK)
display(Markdown("**Raw decode**"))
print(out["text"] or "_(empty)_")
display(Markdown(f"**Verdict** `{out['verdict']}`"))
display(Markdown("**Parsed tags**"))
for k, v in (out.get("parts") or {}).items():
    print(f"{k} {v}")
display(Markdown("**Action**"))
print(out["action"])
```
